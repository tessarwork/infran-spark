```python
import pandas as pd

# Load the CSV file
file_path = 'matched_results_euclidean_parallel.csv'
df = pd.read_csv(file_path)

# Display the first few rows of the dataframe to understand its structure
df.head()
```

```python
import numpy as np

# Define thresholds for similarity (from 0.0 to 1.0 with step size of 0.01)
thresholds = np.arange(0, 1.01, 0.01)

# Assuming that 'Person 1' and 'Person 2' being the same indicates a match (label = 1) and different persons indicate no match (label = 0)
labels = (df['Person 1'] == df['Person 2']).astype(int)

# Similarity scores
scores = 1 - df['Distance'].values

# Lists to store the results
tpr_list = []
fpr_list = []
fnr_list = []  # False Negative Rate (1 - TPR)
far_list = []  # False Acceptance Rate (similar to FPR)

# Calculate TPR, FPR, FNR, and FAR for each threshold
for threshold in thresholds:
    tp = np.sum((scores >= threshold) & (labels == 1))
    fp = np.sum((scores >= threshold) & (labels == 0))
    fn = np.sum((scores < threshold) & (labels == 1))
    tn = np.sum((scores < threshold) & (labels == 0))
    
    # Calculate rates
    tpr = tp / (tp + fn) if (tp + fn) > 0 else 0  # True Positive Rate (TPR)
    fpr = fp / (fp + tn) if (fp + tn) > 0 else 0  # False Positive Rate (FPR)
    fnr = 1 - tpr  # False Negative Rate (FNR)
    far = fpr      # False Acceptance Rate (FAR)
    
    # Append the results
    tpr_list.append(tpr)
    fpr_list.append(fpr)
    fnr_list.append(fnr)
    far_list.append(far)

# Create a DataFrame to display the results
results_df = pd.DataFrame({
    'Threshold': thresholds,
    'TPR': tpr_list,
    'FPR': fpr_list,
    'FNR': fnr_list,
    'FAR': far_list
})

# Set display options and show all rows
pd.set_option('display.max_rows', None)
pd.set_option('display.float_format', '{:.4f}'.format)

# Display the DataFrame to the user
# !pip install ace_tools
# import ace_tools as tools; tools.display_dataframe_to_user(name="Threshold, TPR, FPR, FNR, and FAR Results", dataframe=results_df)

# Display the DataFrame
results_df
```

```python
# Calculate Youden's J statistic
youden_j = np.array(tpr_list) - np.array(fpr_list)

# Find the index of the maximum Youden's J statistic
optimal_index = np.argmax(youden_j)
optimal_threshold = thresholds[optimal_index]

# Get the optimal TPR, FPR, FNR, and FAR at the optimal threshold
optimal_tpr = tpr_list[optimal_index]
optimal_fpr = fpr_list[optimal_index]
optimal_fnr = fnr_list[optimal_index]
optimal_far = far_list[optimal_index]

# Display the optimal values
print(f"Optimal Threshold: {optimal_threshold:.4f}")
print(f"Optimal TPR: {optimal_tpr:.4f}")
print(f"Optimal FPR: {optimal_fpr:.4f}")
print(f"Optimal FNR: {optimal_fnr:.4f}")
print(f"Optimal FAR: {optimal_far:.4f}")
```
